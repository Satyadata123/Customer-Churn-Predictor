### categorical data analyis

```python
# geography
print(df['Geography'].value_counts())
print("\n\n")

result = pd.crosstab(df['Geography'], df['Exited'])
print(result)

sns.catplot(data = df, x = 'Geography', kind = 'count')
```

```python
# gender
print(df['Gender'].value_counts())
print("\n\n")

result = pd.crosstab(df['Gender'], df['Exited'])
print(result)

sns.catplot(data = df, x = 'Gender', kind = 'count')
```

```python
# Tenure
print(df['Tenure'].value_counts())
print("\n\n")

result = pd.crosstab(df['Tenure'], df['Exited'])
print(result)

sns.catplot(data = df, x = 'Tenure', kind = 'count')
```

```python
# NumOfProducts
print(df['NumOfProducts'].value_counts())
print("\n\n")

result = pd.crosstab(df['NumOfProducts'], df['Exited'])
print(result)

sns.catplot(data = df, x = 'NumOfProducts',kind = 'count')
```

```python
# HasCrCard	
print(df['HasCrCard'].value_counts())
print("\n\n")

result = pd.crosstab(df['HasCrCard'], df['Exited'])
print(result)
sns.catplot(data = df , x = "HasCrCard", kind = 'count')
```

```python
# IsActiveMember	
sns.catplot(data = df, x = 'IsActiveMember', kind = 'count')
```

### Numerical data Analysis

```python
# Age 
print(df['Age'].skew())
sns.displot(data = df , x = 'Age', kde = True)
```

```python
# Balance
print(df['Balance'].skew())
sns.displot(data = df, x = 'Balance', kde = True )
```

```python
#EstimatedSalary
print(df['EstimatedSalary'].skew())
sns.displot(data = df, x = 'EstimatedSalary', kde = True)
```
